```python
%load_ext autoreload
%autoreload 2
from pathlib import Path

import magsim.mumax3.cmd as cmd
import magsim.mumax3.blocks as blocks
import magsim.mumax3.script as script
import magsim.mumax3.layer as layer
import magsim.mumax3.material as material
import magsim.mumax3.grid as grid
import magsim.mumax3.helper as helper
```

```python
commands = []
dirpath = Path(r"C:\Users\qqhx82\OneDrive - Durham University\PhD Files\Simulations")

mat = material.permalloy

layers = layer.MagneticLayer(15.0e-9, mat) + layer.empty_layer(2.5e-9) + layer.MagneticLayer(15.0e-9, mat)
layers.assign_dz(2.5e-9)
# gd = grid.Grid(256, 256, 6, 2.5e-9, 2.5e-9, 2.5e-9, (0, 0, 0), 8)
gd = grid.Grid.from_mask(dirpath / "470nm_dot.png", 0.5e-9, 2.5e-9, 2.5e-9, (0, 0, 0), 8)

commands += gd.to_script(*layers.get_nz_and_dz())

commands += layers.to_script()
commands += blocks.load_magstate(
    dirpath,
    index=100,
)

commands += blocks.hysteresis_block(-0.3, 0.3, 1e-3, phi=0.0, commands=[cmd.save("m")])


scr = script.Script(commands)
scr.print()
```
